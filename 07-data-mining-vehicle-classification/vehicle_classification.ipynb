```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.impute import SimpleImputer
```

Load your vehicle dataset 

```python
data = pd.read_csv('vehicles.csv')
```

Explore data overview

```python
print(data.head())
print(data.info())
```

Handle missing values


Impute numerical features with median

```python
data['price'].fillna(data['price'].median(), inplace=True)
data['odometer'].fillna(data['odometer'].median(), inplace=True)
```

Encode categorical features using LabelEncoder

```python
le = LabelEncoder()
categorical_features = ['manufacturer', 'condition', 'fuel', 'title_status', 'transmission', 'drive', 'type', 'paint_color']
for feature in categorical_features:
  data[feature] = le.fit_transform(data[feature])
```

Feature scaling for numerical features

```python
numerical_features = ['price', 'year', 'odometer']  # Removed 'cylinders' from scaling
scaler = StandardScaler()
X = data[numerical_features]
X_scaled = scaler.fit_transform(X)
```

Split data into training and testing sets

```python
X_train, X_test, y_train, y_test = train_test_split(X_scaled, data['type'], test_size=0.2, random_state=42)
```

Impute missing values in numerical features with mean

```python
imputer = SimpleImputer(strategy='mean')
X_train = imputer.fit_transform(X_train)
X_test = imputer.transform(X_test)  # Impute missing values in testing set as well
```

1. Logistic Regression Model

```python
model = LogisticRegression(multi_class='ovr', solver='lbfgs')
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average='weighted')
recall = recall_score(y_test, y_pred, average='weighted')
f1 = f1_score(y_test, y_pred, average='weighted')
```

Logistic Regression Results

```python
print("Logistic Regression Results:")
print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1-Score: {f1:.4f}")
```

2. Decision Tree Model

```python
model = DecisionTreeClassifier()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average='weighted')
recall = recall_score(y_test, y_pred, average='weighted')
f1 = f1_score(y_test, y_pred, average='weighted')
```

Decision Tree Results

```python
print("Decision Tree Results:")
print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1-Score: {f1:.4f}")
```

3. Naive Bayes Model

```python
model = GaussianNB()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average='weighted')
recall = recall_score(y_test, y_pred, average='weighted')
f1 = f1_score(y_test, y_pred, average='weighted')
```

Naive Bayes Results

```python

print("Naive Bayes Results:")
print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1-Score: {f1:.4f}")
```
